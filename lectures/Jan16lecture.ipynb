# Confidence Intervals

## Motivation
How much credence should we give to a statistic from data?

Generate 10, i.i.d. $N(0,1)$ variables, and compute
$$
\bar{x}_n = \frac{1}{n}\sum_{i=1}^n x_i
$$

```python
rng  = np.random.default_rng(seed=123)
x =rng.normal(size=10)
x.mean()
```

Try increasing the sample size:

```python
rng  = np.random.default_rng(seed=123)
x =rng.normal(size=10**6)
x.mean()
```

Need `scipy.stats` to get the values for:
$$
\text{1-$\alpha$ \% Confidence Interval: }\left(\bar{x} - z_{1-\alpha/2}\frac{\hat{\sigma}}{\sqrt{n}},\bar{x} +z_{1-\alpha/2}\frac{\hat{\sigma}}{\sqrt{n}}\right)
$$

```python
from scipy import stats
```

```python
stats.norm.ppf(0.05/2)
```

```python
stats.norm.ppf(1-0.05/2)
```

This the 1.96 for the 95%  confidence interval

More systematically:

```python
sample_mean = x.mean()
sample_se = x.std()/np.sqrt(len(x))
ci_level = 0.95 # 95% confidence interval
ci = stats.norm.interval(ci_level, loc=sample_mean, scale=sample_se)
ci
```

NOTES:
* `x.std()/np.sqrt(len(x))` is the biased estimate of the standard error; divies by `n=len(x)` in computing `x.std()`, rather than `n-1`.  For large `n`, this does _not_ matter.  But to correct it we can do:

```python
x.std(ddof=1)/np.sqrt(len(x)) # computes the standard error with an unbiased variance
```

OR

```python
stats.sem(x) # computes standard error in one shot, using unbiased variance by default
```

To account for finite sample size `n`, use the $t$-distribution instead of $N(0,1)$:

```python
sample_mean = x.mean()
sample_se = stats.sem(x) 
ci_level = 0.95 # 95% confidence interval
ci = stats.t.interval(ci_level, df=len(x)-1, loc=sample_mean, scale=sample_se)
ci
```

For all practical purpsoses, once $n> 30$, we can just use $N(0,1)$
