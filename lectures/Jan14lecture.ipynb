## Math 318 Style Plotting
The textbook heavily encourages the following approach, and it has significant advantages when making composite figures.

### Figure and Axes

````python
# start with
fig, ax = plt.subplots()
````

Creates two objects, `fig` and `ax`.  `fig` is the **entire** figure.  `ax` is the set of axes.  This example only has one set of axes, so `ax` is a single element.

For the boxplot example:

````python
# start with
fig, ax = plt.subplots()
ax.boxplot(x)
ax.set_title("Boxplot of x")
ax.set_xlabel("Case 1")
ax.set_ylabel("Values")
````

So we do `ax.set_xlabel` instead of `plt.xlabel` and so on.

````python
# start with
fig, ax = plt.subplots()
ax.hist(x)
ax.set_title("Histogram of x")
ax.set_xlabel("Values")
ax.set_ylabel("Count")
````

### More Sophitiscated Boxplots

````python
Auto = pd.read_csv("../data/Auto.csv")
Auto.set_index("name", inplace=True)
Auto.cylinders = Auto.cylinders.astype("category")
Auto.origin = Auto.origin.astype("category")
````

Want to viusalize the distribution of mpg, via boxplots, by the cylinders.

First, filter the data we want:

````python
mpg_by_cyl = [Auto[Auto.cylinders == cyl]["mpg"] for cyl in Auto.cylinders.cat.categories]
# looping through the categories of cylinders
# using boolean indexing to select cases with that value of cyl, Auto[Auto.cylinders == cyl]

mpg_by_cyl # this is an array of data frames for thge different cylinder cases
````

````python
fig, ax = plt.subplots()
ax.boxplot(mpg_by_cyl, labels=Auto.cylinders.cat.categories)
ax.set_xlabel("Number of Cylinders")
ax.set_ylabel("Miles Per Gallon")
ax.set_title("MPG by Number of Cylinders")
````

Fast way to do this with `pandas`:

````python
fig, ax = plt.subplots()
Auto.boxplot(column="mpg", by="cylinders", ax=ax)
````

````python
fig, ax = plt.subplots()
Auto.boxplot(column="mpg", by="origin", ax=ax)
````

````python
fig, ax = plt.subplots()
Auto.hist(column="mpg", by="origin", ax=ax)
````

### Using Subplots

````python
# start with
fig, ax = plt.subplots(2,1)
````

This has two axes, `ax[0]` and `ax[1]`, and the single figure, `fig` for the entire image.

In general, we will call
```
fig, ax = plt.subplots(nrow, ncol, figsize=(width, height))
```
Produces a figure with `nrow` by `ncol` subplots, and the full size figure is width by height, measured in inches.

````python
# fig, ax = plt.subplots(2,2)
fig, ax = plt.subplots(2,2, figsize= (10,10))
x = np.linspace(0, 2*np.pi, 100)

ax[0,0].plot(x, np.sin(x))
ax[0,1].plot(x, np.cos(x))
ax[1,0].plot(x, np.sin(2*x))
ax[1,1].plot(x, np.cos(2*x))
# need to annotate each subplot separately
ax[0,0].set_ylabel(r"$\sin(x)$")
fig.savefig("sin_cos.png")
````

# Covariances and Correlations
Recall that
$$
\mathrm{Cov}(X,Y) = \mathbb{E}[(X-\mathbb{E}X)(Y-\mathbb{E}Y)]
$$
and
$$
\mathrm{Corr}(X,Y) = \frac{\mathrm{Cov}(X,Y)}{\sqrt{\mathrm{Var}(X)}\sqrt{\mathrm{Var}(Y)}}
$$

````python
Auto.corr() # gives pair correlations amongst features (numeric or otherwise)
````

Filter out the categorical comparison

````python
Auto.corr(numeric_only=True) # gives pair correlations amongst features (numeric  only)
````

Correlation matrices always have:
* Symmetry, $C = C^T$
* Diagonal is ones

Getting the actual matrix:

````python
C = Auto.corr(numeric_only=True).to_numpy()
C
````

````python
Auto[['mpg', 'horsepower', 'weight']].corr(numeric_only=True) # gives pair correlations amongst features (numeric  only)
````
