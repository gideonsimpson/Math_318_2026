```python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd

import statsmodels.api as sm

from ISLP.models import (ModelSpec as MS, summarize , poly)
from ISLP import confusion_table
```

```python
# Load Data
```

```python
default = pd.read_csv("../data/Default.csv")
default['student'] = default['student'].astype('category')
default['default'] = default['default'].astype('category')
default.head()
```

# Test/Train Split

```python
# set seed for reproducibility
train_df = default.sample(frac=0.7, random_state=123)   # 70% train
test_df  = default.drop(train_df.index)                 # 30% test
```

```python
design = MS(['balance'])
X_train = design.fit_transform(train_df)
y_train = train_df['default'] == 'Yes'

X_test = design.transform(test_df)
y_test = test_df['default'] == 'Yes'

glm = sm.GLM(y_train, X_train, family=sm.families.Binomial()) 
results = glm.fit() 
```

## Training Performance

```python
# predicted default probabilities on training data
train_p_pred = results.predict(X_train)

# assign labels to these predictions:
train_labels_pred = np.full(len(train_df), 'No', dtype=object)
train_labels_pred[train_p_pred > 0.5] = 'Yes'

print(f'Training error = {np.mean(train_labels_pred != train_df['default'])}')
```

```python
# predicted default probabilities on testing data
test_p_pred = results.predict(X_test)

# assign labels to these predictions:
test_labels_pred = np.full(len(test_df), 'No', dtype=object)
test_labels_pred[test_p_pred > 0.5] = 'Yes'

print(f'Testing error = {np.mean(test_labels_pred != test_df['default'])}')
```
