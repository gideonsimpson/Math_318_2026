```python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd
```

# Linear Regression with One Feature
Try to fit $(x_i,y_i)_{i=1}^n$ data with
$$
Y \approx \beta_0 + \beta_1 X
$$

## Generate Data
Start with toy data
$$
y_i = 2 - 1.5 x_i + \text{noise}
$$

```python
x_ = np.linspace(0, 10, 15)
rng = np.random.default_rng(1234)
y_ = 2 - 1.5 * x_ + rng.normal(size=x_.shape)
```

```python
plt.scatter(x_, y_)
plt.xlabel('x')
plt.ylabel('y') 
plt.title('Scatter plot of y vs x')
plt.show()
```

Store in a data frame for later:

```python
df = pd.DataFrame({'x': x_, 'y': y_})
df.head()
```

## sklearn
If for some reason we do not wish to use `statsmodels`, we can use the `scikit-learn` equivalent method:

```python
# Ordinary Least Squares (OLS) with scikit-learn
from sklearn.linear_model import LinearRegression
```

`sklearn` expects a data frame for the design matrix (2D array):

```python
X = df[['x']] # shape (n, 1)
# compare with what you get with X = df['x']
X
```

```python
# built the model
lr = LinearRegression(fit_intercept=True) # fit_intercept=True adds the \beta_0 term 
# fit the model
lr.fit(X, df['y'])
```

* Coefficients are stored in `intercept_` and `coef_` elements
* $R^2$ is computed using hte `score` function

```python
print('beta0 = {}'.format(lr.intercept_))
print('beta1 = {}'.format(lr.coef_[0]))
print('R^2:', lr.score(X, df['y']))
```

```python
fig, ax = plt.subplots()
df.plot.scatter(x='x', y='y', ax=ax, label='Data')
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$y$')

# plot the regression line
# this evaluates \hat{y} = \beta_0 + \beta_1 x for convenient x values
yhat = lr.predict(X) 
ax.plot(df['x'], yhat, color='C1', label='OLS fit')
ax.legend()
```

The `'C1'` referes to "Color 1" in the `matplotlib` naming convention.  The default, dark blue, is `'C0'`, and this goes up to `'C9'`.

Overall, the automatically generated information with `sklearn` regression is a bit briefer, but we can obtain additional statistics, like the $F$-statistic and its $p$-value:

```python
# import this for getting the F statistic
from sklearn.feature_selection import f_regression
F_values, p_values = f_regression(X, df['y'])
# note the use of string formatting to control the number of digits displayed
print('F statistic: {:.3f}\np-value: {:.3g}'.format(F_values[0], p_values[0]))
```

Making new predictions with our model is similar:

```python
new_df = pd.DataFrame({'x': [2.23, 4.53, 6.78]})
preds = lr.predict(new_df[['x']])
preds
```

```python
fig, ax = plt.subplots()
df.plot.scatter(x='x', y='y', ax=ax, label='Data')
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$y$')

# plot the regression line
# this evaluates \hat{y} = \beta_0 + \beta_1 x for convenient x values
yhat = lr.predict(X) 
ax.plot(df['x'], yhat, color='C1', label='OLS fit')
ax.scatter(new_df['x'], preds, color='C2', label='New Points')
ax.legend()
```
