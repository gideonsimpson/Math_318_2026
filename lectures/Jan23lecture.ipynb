````python
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd
````

# Linear Regression with One Feature
Try to fit $(x_i,y_i)_{i=1}^n$ data with
$$
Y \approx \beta_0 + \beta_1 X
$$

## Generate Data
Start with toy data
$$
y_i = 2 - 1.5 x_i + \text{noise}
$$

````python
x_ = np.linspace(0, 10, 15)
rng = np.random.default_rng(1234)
y_ = 2 - 1.5 * x_ + rng.normal(size=x_.shape)
````

````python
plt.scatter(x_, y_)
plt.xlabel('x')
plt.ylabel('y') 
plt.title('Scatter plot of y vs x')
plt.show()
````

Store in a data frame for later:

````python
df = pd.DataFrame({'x': x_, 'y': y_})
df.head()
````

## statsmodels

````python
import statsmodels.api as sm
````

Need to construct the design matrix for this approach (i.e., using `statsmodels`).   The design matrix will be:
$$
\mathcal{X} = \begin{pmatrix} 1& x_1\\
1 & x_2\\
\vdots & \vdots\\
1 & x_n
\end{pmatrix}
$$

````python
X = df['x'] # the design matrix
X
````

To get the columns of ones:

````python
X = sm.add_constant(df['x'])  # Adds a constant term to the predictor
X
````

`X` contains the design matrix stored as a data frame

Set up the problem, and then solve:

````python
model = sm.OLS(df['y'], X) # OLS = Ordinary Least Squares, specify the response variable and the design matrix
results = model.fit()  # Fit the model
````

````python
results
````

Get some information out:

````python
results.summary()
````

````python
results.params # pull out just the covariates
````

 * $R^2$ - the fraction of observered variation in $y$ explained by the model.  
 * $\mathrm{Adj.}-R^2$ - compensates for number of features, $p$, in a mdoel.  Here $p=1$
 * $F$-statistic and is probability - tell us the statistical significance of the model.  Comparing the $P(F-\mathrm{statistic})$ agianst $\alpha = 0.05$, for exmaple
 * For each feature, the $t$-statistic and its probability would be compared against an $\alpha$, and null hypothesis is $\beta_j = 0$.

Scratch calculation for looking at two models, with the same $R^2 = .85$, $n=20000$, the first model has $p=1$, and the second model has $p=1000$

````python
R2 = .85
n = 2000
print(1 - (1 - R2)*(n - 1)/(n - 1 - 1)) # model with p=1
print(1 - (1 - R2)*(n - 1)/(n - 1000 - 1)) # model with p =1000
````

Fewer features (simpler) is bettter

Making new predictions the model:

````python
x_new = 0.85
x_new_df = pd.DataFrame({'x': [x_new]})
x_new_df = sm.add_constant(x_new_df, has_constant='add')  # add constant term, need has_constant='add' because there is only 1 new point
y_new_pred = results.predict(x_new_df)
y_new_pred
````

````python
x_new_df = pd.DataFrame({'x': [0.85, 3, 2.5]})
x_new_df = sm.add_constant(x_new_df)  # add constant term
y_new_pred = results.predict(x_new_df)
y_new_pred
````
