## Setting the Seed

For all work involving (pseudo) random number generation, you should explicitly set a _seed_

```python
rng = np.random.default_rng(123) # create a random number generator object, set the seed to 123
```

```python
rng.standard_normal() # samples N(0,1)
```

```python
rng.standard_normal(size=(3,3))
```

```python
rng = np.random.default_rng(123) # rest the seed
```

```python
rng.normal(loc=-1, scale=1, size=(3,3)) # reproducibily sample N(-1,1) random numbers in a 3x3 array
```

How is the variance being computed in NumPy?  First look at a smaller data set:

```python
rng= np.random.default_rng(123) # rest the seed
data = rng.standard_normal(10)
```

By hand:

```python
m = data.mean()
var = (((data - m) ** 2).sum())/(data.size-1)
print(f'm = {m}, var = {var}')
```

```python
data.var()
```

These are not the same

```python
m = data.mean()
var = (((data - m) ** 2).sum())/(data.size)
print(f'm = {m}, var = {var}')
```

The **default** variance command in NumPy divides by $N$, not $N-1$

```python
data.var(ddof=1) # divide by N - ddof (set ddof to 1 to get the sample variance)
```
