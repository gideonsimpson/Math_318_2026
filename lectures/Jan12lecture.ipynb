## If-Then-Else

Note that we have to indent by 1 tab or 4 spaces, for whatever comes within the clause:

# Intro to Plotting

Create some data for visualization

```python
x = np.linspace(-5,5,100)
y = np.sin(x)
```

## Scatter Plots

```python
plt.scatter(x,y)
plt.plot(x,y)
plt.show()
```

Plotting some points and not others; use Boolean indexing

```python
x[y>0]
```

```python
# plt.scatter(x[np.where(y>0)], y[np.where(y>0)], color='red')
plt.scatter(x[y>0], y[y>0], color='red')
plt.plot(x,y)
plt.show()
```

```python
# plt.scatter(x[np.where(y>0)], y[np.where(y>0)], color='red')
plt.scatter(x[(y>0)&(y<.5)], y[(y>0)&(y<.5)], color='red')
plt.plot(x,y)
plt.show()
```

For boolean computations, `&` is the **and** condition, and `|` is the **or** condition:

```python
# plt.scatter(x[np.where(y>0)], y[np.where(y>0)], color='red')
plt.scatter(x[(y>.5)|(y<-.5)], y[(y>.5)|(y<-.5)], color='red')
plt.plot(x,y)
plt.show()
```

## Histograms

```python
# Generate some normal distribution data and plot it
rng = np.random.default_rng(seed=0)
x = rng.normal(size=10000)
plt.hist(x, bins=30)
plt.show()
```

```python
# annotate the figure
plt.hist(x, bins=30, label=r'Data from $N(0,1)$')
plt.xlabel(r'$x$')
plt.ylabel('Count')
plt.legend()
plt.show()
```

```python
# plot in terms of probability density, and add in the density function
xx = np.linspace(-4,4,100)
plt.hist(x, bins=30, label=r'Data from $N(0,1)$', density=True)
plt.plot(xx, 1/np.sqrt(2*np.pi)*np.exp(-0.5*xx**2), label=r'$N(0,1)$ PDF')
plt.xlabel(r'$x$')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
